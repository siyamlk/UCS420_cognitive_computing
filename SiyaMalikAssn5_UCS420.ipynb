#### Q.1 Write a program to create a NumPy 1D-array with 5 elements and perform basic operations like:
##### a) Addition of 2 in all the element
##### b) Multiply 3 with all the elements
##### c) Divide every element by 2

#### Q.2 Questions on Basic NumPy Array:
##### a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5])

```python
arr=np.array([1,2,3,6,4,5])
print("Reversed Array: ",np.flip(arr))
```

##### b) Find the most frequent value and their indice(s) in the following arrays:

```python
x = np.array([1,2,3,4,5,1,2,1,1,1])
vals,counts=np.unique(x,return_counts=True)
most_frequent=vals[np.argmax(counts)]
indices=np.where(x==most_frequent)
print(f"Most frequent element: {most_frequent} ")
print(f"Indices of most frequent element: {indices}")
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])
vals,counts=np.unique(y,return_counts=True)
most_frequent=vals[np.argmax(counts)]
indices=np.where(y==most_frequent)
print("Most frequent element: ",most_frequent)
print("Indices of most frequent element: ",indices)
```

#### Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array, shape, total elements, the data type of each element and total number of bytes consumed by the array. Find the transpose of this array using reshape() attribute. Can we do the same with T attribute?

```python
siya=np.linspace(10,100,25)
print("Array: ",siya)
print("Dimension of the array: ",np.ndim(siya))
print("Shape of the array: ",np.shape(siya))
print("Total elements: ",np.size(siya))
print("Data type of each element: ",siya.dtype)
print("Total Bytes: ",siya.nbytes)

```

```python
print("Reshape using reshape()")
print(siya.reshape(1,25).T)
```

```python
print("Reshape using T")
print(siya.T)
```

##### The T attribute can be used for transposition, but for a 1-D NumPy array, T has no effect. Therefore, the array must first be reshaped into a 2-D array and then transposed.

#### Q5. Create a 2-D Array of three rows and four columns, named ucs420_<\your_name>> with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it as reshaped_ ucs420_<\your_name>>. Resize the array to two rows and three columns and name it as resized_ ucs420_<\your_name>>.

```python
ucs420_siya=np.array(([10,20,30,40],[50,60,70,80],[90,15,20,35]))
print("Array: ",ucs420_siya)
print("Shape of the array: ",np.shape(ucs420_siya))
```

```python
print("Mean: ",np.mean(ucs420_siya))
print("Median: ",np.median(ucs420_siya))
print("Max element: ",np.max(ucs420_siya))
print("Min element: ",np.min(ucs420_siya))
print("Unique elements: ",np.unique(ucs420_siya))
```

```python
reshaped_ucs420_siya=ucs420_siya.reshape((4,3))
print("Reshaped Array: ",reshaped_ucs420_siya)
```

```python
resized_ucs420_siya=np.resize(ucs420_siya,(2,3))
print("Resized Array: ",resized_ucs420_siya)
```
