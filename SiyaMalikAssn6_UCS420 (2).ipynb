#### Q1. Consider the following sensor readings: temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40]) 
#### A sensor has an average measurement error of +2°C. Perform the following without using a for loop: 

##### a. Add 2°C to every reading using vectorization. 

```python
import numpy as np
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
temperature_corrected = temperature + 2
print(temperature_corrected)
```

##### b. Convert all temperatures from Celsius to Fahrenheit

```python
fahrenheit = (9/5) * temperature_corrected + 32
print(fahrenheit)
```

##### c. Identify all readings greater than 32°C using Boolean indexing. 

```python
readings = temperature_corrected[temperature_corrected > 32]
print(readings)
```

##### d. Count how many readings exceed 32°C. 

```python
count = np.sum(temperature_corrected > 32)
print(count)
```

##### e. Explain how vectorization and Boolean indexing can make data processing more efficient than explicitly iterating through every value. 

Vectorization makes data processing efficient by performing operations on an entire array at once instead of using loops to process each element individually. This reduces execution time and makes the code shorter and easier to understand.

Boolean indexing allows us to filter and select only the elements that satisfy a given condition. For example, it can directly select all temperature readings greater than 32°C without checking each value separately.

#### Q4. A simple cognitive-assistive system wants to predict a user's assistance score based on three factors: • Sleep hours  • Activity level  • Stress level  Consider: y = np.array([40, 65, 30, 85]) X = np.array([ [6, 70, 3], [5, 50, 6], [8, 80, 2], [4, 30, 8] ]) Here, each row represents one user and each column represents a feature. Perform the following: 

##### a. Display the shape and dimensions of X. 

```python
import numpy as np
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])
print(X.shape)
print(X.ndim)
```

##### b. Calculate: X.T and explain what the transpose represents.

```python
XT = X.T
print(XT)
```

##### c. Calculate the matrix product: X.T @ X 

```python
product = X.T @ X
print(product)
```

##### d. Calculate the matrix inverse using: np.linalg.inv() where appropriate. 

```python
inverse = np.linalg.inv(X.T @ X)
print(inverse)
```

##### e. Implement the Ordinary Least Squares equation: 𝛽 = (𝑋𝑋) 𝑋𝑦 using NumPy.

```python
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print(beta)
```

##### f. Explain what the resulting coefficients represent in the context of the three user features. 

The coefficients represent the effect of each feature on the assistance score, while keeping the other features constant.

Sleep hours coefficient = -0.1592: A 1-unit increase in sleep hours decreases the predicted assistance score by about 0.1592. 
Activity level coefficient = 0.1408: A 1-unit increase in activity level increases the predicted assistance score by about 0.1408. 
Stress level coefficient = 10.0630: A 1-unit increase in stress level increases the predicted assistance score by about 10.0630. 

Thus, the coefficients indicate how strongly each of the three features influences the predicted assistance score.

##### g. A new user has: new_user = np.array([5, 40, 7]) Use your calculated coefficients to obtain a predicted assistance score. 

```python
new_user = np.array([5, 40, 7])
prediction = new_user @ beta
print(prediction)
```
