```python
n = int(input("Enter a number: "))

sum = 0

for num in range(2, n + 1):
    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        sum += num

print("Sum of prime numbers =", sum)
```

```python
def IsPrime(n):
	for i in range(2, n//2 + 1):
		if n%i==0:
			return 0
	return 1

print ("IsPrime(20)  --> ", IsPrime(27))
print ("IsPrime(23)  --> ", IsPrime(293))
print ("IsPrime(200) --> ", IsPrime(257))
print ("IsPrime(37)  --> ", IsPrime(89))
```

```python
def AddOdd(n):
    total = 0
    for i in range(1, n + 1):
        if i % 2 != 0:
            total += i
    return total

n = int(input("Enter a number: "))
print("Sum of odd numbers =", AddOdd(n))
```

```python
def AddPrime(n):
    total = 0

    for num in range(2, n + 1):
        prime = True

        for i in range(2, num):
            if num % i == 0:
                prime = False
                break

        if prime:
            total += num

    return total

n = int(input("Enter a number: "))
print("Sum of prime numbers =", AddPrime(n))
```

```python
import string as s
import random as r

chars = s.ascii_lowercase + s.digits
print("Characters -->", chars)

pwd1 = r.sample(chars, 6)
print("\nSelected Password 1 -->", pwd1)
pwd1 = "".join(pwd1)
print("Password 1 -->", pwd1)

pwd2 = r.sample(chars, 8)
print("\nSelected Password 2 -->", pwd2)
pwd2 = "".join(pwd2)
print("Password 2 -->", pwd2)

special = "@#$%&*!"
pwd3 = r.sample(chars + special, 10)
print("\nSelected Password 3 -->", pwd3)
pwd3 = "".join(pwd3)
print("Password 3 -->", pwd3)
```
