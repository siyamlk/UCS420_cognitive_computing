```python
def convert_rollnumber_to_list(rollnumber):
    result=[]
    for i in rollnumber:
        if(i.isdigit()):
            digit=int(i)
            newDigit=digit*10
            result.append(newDigit)
    return result
```

```python
result=convert_rollnumber_to_list("1024170405")
```

```python
import random
random.seed(1024170405)
numbers = [random.randint(100, 900) for _ in range(100)]
print(numbers)
```

```python
odd_count = 0
for num in numbers:
    if num % 2 != 0:
        odd_count += 1
print("Number of odd numbers:", odd_count)
```

```python
even_count = 0
for num in numbers:
    if num % 2 == 0:
        even_count += 1
print("Number of even numbers:", even_count)
```

```python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True
```

```python
prime_numbers = [num for num in numbers if is_prime(num)]
print("Number of prime numbers:", len(prime_numbers))
print("Prime numbers:", prime_numbers)
```

```python
most_frequent = max(numbers, key=numbers.count)
frequency = numbers.count(most_frequent)
print("Most frequent number:", most_frequent)
print("Number of times it occurs:", frequency)
```

```python
digits = [1, 0, 2, 4, 1, 7, 0, 4, 0, 5]
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}
print("A:", A)
print("B:", B)
```

```python
union = A.union(B)
print("Union:", union)
```

```python
intersection = A.intersection(B)
print("Intersection:", intersection)
```

```python
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)
print("A - B:", A_minus_B)
print("B - A:", B_minus_A)
```

```python
sym_diff = A.symmetric_difference(B)
print("Symmetric difference:", sym_diff)
```

```python
print("A is subset of B:", A.issubset(B))
print("B is superset of A:", B.issuperset(A))
```

```python
X = int(input("Enter a value: "))
A.discard(X)
print("Updated A:", A)
```
